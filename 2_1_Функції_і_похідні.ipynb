1. Задана функція залежності оцінки від кількості годин, витрачених на навчання: $ f(x) = 10x - 5 $.
  1. З допомогою Python обчисліть похідну цієї функції.
  2. Використовуйте бібліотеку `matplotlib` для побудови графіків функції та її похідної в діапазоні значень від 0 до 10 годин. Проаналізуйте, як зміна кількості годин вивчення впливає на оцінку.



```python
import sympy as sp
import matplotlib.pyplot as plt
import numpy as np

x = sp.symbols('x')
f = 10*x - 5
df = sp.diff(f, x)

print('differential of 10*x - 5 is', df)
```

```python
x = np.linspace(0, 10, 100)
f = 10*x - 5
df = np.full_like(x, 10)

#plot
plt.figure(figsize=(10, 5))
plt.plot(x, f, label="f(x)")
plt.plot(x, df, label="f'(x)")

plt.xlabel('Кількість годин (x)')
plt.ylabel('Оцінка / Похідна')
plt.legend()
plt.grid(True)

plt.show()
```

**Висновки**

Чим більше годин витрачено на навчання, тим вище оцінка

Похідна - константа(від лінійної функції),
кожна додаткова година навчання підвищує оцінку на 10 балів

2. Використовуючи Python і `SymPy`, обчисліть похідні для функцій $ f(x) = \sin(x) $ та $ f(x) = \ln(x) $. Зобразіть ці функції та їхні похідні за допомогою `matplotlib`.



```python
x = sp.symbols('x')
fx = sp.sin(x)
df = sp.diff(fx, x)

print('differential of sin(x) is', df)
```

```python
x = np.linspace(0, 10, 100)
fx = np.sin(x)
df = np.cos(x)

#plot
plt.figure(figsize=(10, 5))
plt.plot(x, fx, label="f(x) = sin(x)")
plt.plot(x, df, label="f'(x) = cos(x)")

plt.legend()
plt.grid(True)

plt.show()
```

```python
x = sp.symbols('x')
fx = sp.ln(x)
df = sp.diff(fx, x)

print('differential of ln(x) is', df)
```

```python
x = np.linspace(0.1, 10, 100)
fx = np.log(x)
df = 1/x

#plot
plt.figure(figsize=(10, 5))
plt.plot(x, fx, label="f(x) = ln(x)")
plt.plot(x, df, label="f'(x) = 1/x")

plt.legend()
plt.grid(True)

plt.show()
```


3. Загальні витрати компанії від обсягу виробництва задані функцією $ f(x) = x^2 - 5x + 1 $, де $ x $ — кількість вироблених одиниць. Обчисліть похідну цієї функції з допомогою SymPy та візуалізуйте значення функції і її похідної на відрізку від 0 до 100.

```python
x = sp.symbols('x')
fx = x**2 - 5*x + 1
df = sp.diff(fx, x)

print('differential of x**2 - 5*x + 1 is', df)
```

```python
x = np.linspace(0,100, 1000)
fx = x**2 - 5*x + 1
df = 2*x -5

plt.figure(figsize=(10,6))
plt.plot(x, fx, label = "f(x) = x**2 - 5*x + 1")
plt.plot(x, df, label = "f'(x) = 2*x - 5")

plt.xlabel('Одиниці продукції (x)')
plt.ylabel('Витрати / Похідна')
plt.legend()
plt.grid(True)

plt.show()
```
