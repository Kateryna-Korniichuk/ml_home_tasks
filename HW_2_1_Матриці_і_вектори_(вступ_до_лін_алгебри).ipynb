
### Завдання 3: Множення матриць
1. Задані дві матриці `A` і `B`розмірностями 3x2 та 2х3. Виконайте їхнє множення, щоб отримати матрицю 2x2. Виведіть результуючу матрицю `result_matrix` на екран.

2. Заданий вектор `C` розмірністю 2х1. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на вектор `C` і який результат буде. А потім виконайте множення і виведіть результат на екран.

3. Задана матриця `D` розмірністю 2х2. Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите `result_matrix` на матрицю `D` і який результат буде. А потім виконайте множення і виведіть результат на екран.

```python
A = np.array(
    [[0.18, 0.53],
     [0.75, 0.47],
     [0.35, 0.21]])

B = np.array(
    [[0.17, 0.58, 0.75],
     [0.38, 0.11, 0.15]])

C = np.array([[0], [1]])

D = np.array([[0,1], [1,0]])

display(A, B, C, D)
```

```python
# Задані дві матриці A і Bрозмірностями 3x2 та 2х3.
# Виконайте їхнє множення, щоб отримати матрицю 2x2.
# Виведіть результуючу матрицю result_matrix на екран.

result_matrix = np.dot(B, A)

result_matrix
```

```python
# Заданий вектор C розмірністю 2х1.
# Подумайте спочатку, яку розмірність Ви отрмаєте,
# коли домножите result_matrix на вектор C і який результат буде.
# А потім виконайте множення і виведіть результат на екран.

np.dot(result_matrix, C)
```

множення на вектор (0, 1) фактично залишило нам тільки другий стовпчик

```python
# Задана матриця D розмірністю 2х2.
# Подумайте спочатку, яку розмірність Ви отрмаєте, коли домножите
# result_matrix на матрицю D і який результат буде.
# А потім виконайте множення і виведіть результат на екран.

np.dot(result_matrix, D)
```

а в цьому заданні стовпчики помінялись місцями

```python
#Транспонуйте result_matrix.

result_matrix_T = result_matrix.T

result_matrix_T
```

```python
# Зробіть транспонування вектора vector і виведіть форму нової структури. Чи бачите ви зміну?

vector_T = vector.T

print(vector_T)
print(vector_T.shape)
```

ні, зміни немає, бо це одновимірний вектор

```python
# Використайте операцію numpy.expand_dims над вектором vector з аргументом axis=1.
# Запишіть отриманий результат в змінну column_vector і виведіть на екран. Порівняйте з vector.

column_vector = np.expand_dims(vector, axis=1)

print(vector, vector.shape)
print(column_vector, column_vector.shape)
```

vector - одновимірний вектор

column_vector - матриця розмірности (10,1)

```python
# Зробіть транспонування column_vector, виведіть результат на екран разом з розмірностями отриманої структури.

print(column_vector.T)
print(column_vector.T.shape)
```
